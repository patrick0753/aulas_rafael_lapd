# Criação do banco de dados
- Importação de bibliotecas

```python
import sqlite3
import pandas as pd
```

# Criando conexão com o banco

```python
# Cria (ou abre, caso já exista) o banco de dados "empresa.db"
conn = sqlite3.connect("empresa.db")

# O objeto cursor é quem efetivamente executa os comandos SQL
cursor = conn.cursor()
```

# Criando tabela 

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS funcionarios (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nome TEXT NOT NULL,
    cargo TEXT NOT NULL,
    salario REAL NOT NULL,
    data_contratacao TEXT NOT NULL);
""")
print('tabela criada com sucesso')
```

# Informações da tabela - Metadados

# Consultando os registros (SELECT)

```python
cursor.execute("SELECT * FROM funcionarios;")
todos = cursor.fetchall()

print("Todos os funcionários cadastados:\n")
for funcionario in todos:
    print(funcionario)
```

# Utilizando WHERE

```python
cursor.execute("SELECT * FROM funcionarios WHERE salario > 4500;")
salarios_altos = cursor.fetchall()

print("Funcionários com salário acima de R$ 4500:\n ")
for i in salarios_altos:
    print(i)
```

```python
# funcionarios com cargo avançado kkkkkkkkkk
cursor.execute("SELECT * FROM funcionarios WHERE cargo = 'Caixa de Supermercado';" )
caixinha = cursor.fetchall()

print("Funcionários com profissão boa: \n")
for i in caixinha:
    print(i)
```
