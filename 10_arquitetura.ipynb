### DADOS EM TRÂNSITO E REPOUSO

**Em trânsito**: O dado circula entre componentes por uma rede ou por comunicação entre processos -> criptografia em trânsito
***Analogia*** É uma encomenda no caminhão


**Em repouso**: O dado permanece gravado em arquivo, banco, objeto ou outro meio persistente -> acesso, backup, criptografia
***Analogia***: É uma encomenda no depósito
