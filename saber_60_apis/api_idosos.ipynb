### Bibliotecas

```python
import requests
import pandas as pd
import time
```

### Utilizamos SIDRA IBGE
- Utilizaremos a tabela 9514

```python
url = "https://servicodados.ibge.gov.br/api/v3/agregados/9514/periodos/2022/variaveis/93?localidades=N3[35]"
print("enviando requisicao...")
resposta = requests.get(url)
status = resposta.status_code
if status == 200:
    print("requisição bem sucedida")
elif status == 404:
    print("recurso não encontrado")
else:
    print(f"erro desconhecido: {status}")
```
